#***5 — Feature Engineering & Data Pre-processing.***

## Feature Engineering and Data Pre-processing

### Encode the Target Variable

Machine Learning classification models generally require the target categories to be represented numerically. `LabelEncoder` converts the species labels into numerical class values.

The encoded values represent the three flower species. The `value_counts()` output shows the number of observations belonging to each encoded class after duplicate removal.

```python
# Encode the species column

le = LabelEncoder()

data['species'] = le.fit_transform(data['species'])

data['species'].value_counts()
```

### Separate Features and Target

The input features (`X`) are the four flower measurements: sepal length, sepal width, petal length, and petal width.

The target variable (`y`) is the encoded species column. The model uses the four measurements to predict the flower's species.

```python
# Separate Features (X) and Target (y)

X = data.drop('species', axis=1)
y = data['species']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

### Split the Dataset into Training and Testing Sets

The dataset is divided into training and testing sets. The training set contains 70% of the observations, while the testing set contains 30%.

The `stratify=y` parameter helps preserve the species distribution across both sets. The `random_state=42` parameter makes the split reproducible.

In this dataset, 102 observations are used for training and 45 for testing.

```python

# Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

## ***6. ML Model Implementation***

## Machine Learning Model Implementation

### Define Classification Models

In this section, seven Machine Learning classification algorithms are defined:

- Logistic Regression
- Decision Tree
- Random Forest
- K-Nearest Neighbors (KNN)
- Linear Discriminant Analysis (LDA)
- Gaussian Naive Bayes
- Support Vector Machine (SVM)

These models will be trained on the same training dataset and evaluated on the same test dataset so their classification accuracy can be compared.

```python

# Define Machine Learning models

models = {
    'Logistic Regression': LogisticRegression(max_iter=200),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'K-Nearest Neighbors': KNeighborsClassifier(),
    'Linear Discriminant Analysis': LinearDiscriminantAnalysis(),
    'Naive Bayes': GaussianNB(),
    'Support Vector Machine': SVC()
}

print("Total models:", len(models))
```

### Train and Evaluate the Models

Each model is trained using the training data and then used to predict the species in the test data.

The `accuracy_score()` function calculates the proportion of correct predictions. The results are stored in a DataFrame and sorted by accuracy to compare model performance.

The results show that **K-Nearest Neighbors and Linear Discriminant Analysis achieve the highest test accuracy of 97.78%**.

```python

# Train and evaluate each model

results = []

for name, model in models.items():
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    results.append({
        'Model': name,
        'Accuracy': accuracy
    })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(
    by='Accuracy',
    ascending=False
)

print(results_df.to_string(index=False))
```

### Select and Evaluate the Best-Performing Model

The model with the highest test accuracy is selected from the comparison results. In this experiment, K-Nearest Neighbors is selected; it ties with Linear Discriminant Analysis at 97.78% accuracy.

The classification report provides precision, recall, F1-score, and support for each species. These metrics give a more detailed view of classification performance than accuracy alone.

```python
# Select the model with the highest test accuracy

best_model_name = results_df.iloc[0]['Model']
best_model = models[best_model_name]

y_pred = best_model.predict(X_test)

print("Best Model:", best_model_name)
print("\nClassification Report:")

# Cast class names to strings to avoid the TypeError
target_names = [str(c) for c in le.classes_]
print(classification_report(y_test, y_pred, target_names=target_names))
```
